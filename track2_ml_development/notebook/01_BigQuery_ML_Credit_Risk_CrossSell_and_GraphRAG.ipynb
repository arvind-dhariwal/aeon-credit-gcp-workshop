## Step 2 (Backup Code — Module 1): Governed Feature Store + Property Graph Contagion Features (`acsm_gold.ml_customer_feature_store`)
**ACSM RFP Clause**: `C1.1.4.2` *(Feature Engineering & Governed Feature Store)*

Before training our models, **Module 1** builds `acsm_gold.ml_customer_feature_store` across all `100,000` customers by combining:
1. **Customer 360 Tabular Features (`acsm_gold.gold_aeon_customer360_profile`)**: Demographics (`N_Age`, `State`, `Region`, `Occupation`), Income (`B_NetIncome`, `B_AnnualIncome`), Easy Payment (`EP`) underwriting (`ep_app_count`, `total_ep_financed_myr`, `avg_ep_dsr`), Credit Card (`CC`) underwriting (`cc_app_count`, `total_cc_limit_myr`, `latest_ctos_score`), and Card utilization (`active_card_count`, `credit_utilization_ratio`).
2. **BigQuery Property Graph Contagion Features (`GRAPH_TABLE` over `acsm_gold.acsm_credit_ecosystem_graph`)**: Traverses `(c:Customer)-[:TRANSACTED_AT]->(m:Merchant)` to calculate how many **high-delinquency merchants** (`delinquent_exposed_merchants`) and **shared delinquent peers** (`shared_delinquent_peers_count`) each customer is connected to in the graph!

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 1.1: Build Governed Feature Store (`acsm_gold.ml_customer_feature_store`)
-- Joining Customer 360 features with BigQuery Property Graph (`GRAPH_TABLE`) merchant contagion features
CREATE OR REPLACE TABLE `acsm_gold.ml_customer_feature_store`
CLUSTER BY State, CIF_ID
OPTIONS (
  description = 'Governed ACSM ML Feature Store combining Customer 360 demographics, EP/CC underwriting, DSR, CTOS score, and Property Graph merchant contagion features (Clause C1.1.4.2).'
) AS
WITH high_risk_merchants AS (
  SELECT
    merchant_id,
    COUNT(DISTINCT delinq_cif_id) AS delinquent_shoppers_at_merchant
  FROM GRAPH_TABLE(
    `acsm_gold.acsm_credit_ecosystem_graph`
    MATCH (dc:Customer)-[t:TRANSACTED_AT]->(m:Merchant)
    WHERE dc.is_delinquent = TRUE
    COLUMNS (
      m.merchant_id AS merchant_id,
      dc.cif_id AS delinq_cif_id
    )
  )
  GROUP BY merchant_id
  HAVING COUNT(DISTINCT delinq_cif_id) >= 5
),
customer_graph_contagion AS (
  SELECT
    g.cif_id,
    COUNT(DISTINCT hrm.merchant_id) AS delinquent_exposed_merchants,
    COALESCE(SUM(hrm.delinquent_shoppers_at_merchant), 0) AS shared_delinquent_peers_count,
    ROUND(SUM(g.total_spend_myr), 2) AS total_graph_merchant_spend_myr
  FROM GRAPH_TABLE(
    `acsm_gold.acsm_credit_ecosystem_graph`
    MATCH (c:Customer)-[t:TRANSACTED_AT]->(m:Merchant)
    COLUMNS (
      c.cif_id AS cif_id,
      m.merchant_id AS merchant_id,
      t.total_spend_myr AS total_spend_myr
    )
  ) g
  LEFT JOIN high_risk_merchants hrm
    ON g.merchant_id = hrm.merchant_id
  GROUP BY g.cif_id
)
SELECT
  p.CIF_ID,
  p.CIF_NM,
  COALESCE(p.State, 'Unknown') AS State,
  COALESCE(p.Region, 'Unknown') AS Region,
  COALESCE(p.Occupation, 'Unknown') AS Occupation,
  CAST(COALESCE(p.N_Age, 35) AS INT64) AS N_Age,
  CAST(COALESCE(p.B_NetIncome, 0) AS FLOAT64) AS B_NetIncome,
  CAST(COALESCE(p.B_AnnualIncome, 0) AS FLOAT64) AS B_AnnualIncome,
  COALESCE(p.RecvPromo_FG, 'N') AS RecvPromo_FG,
  CAST(COALESCE(p.ep_app_count, 0) AS INT64) AS ep_app_count,
  CAST(COALESCE(p.total_ep_financed_myr, 0) AS FLOAT64) AS total_ep_financed_myr,
  CAST(COALESCE(p.avg_ep_dsr, 30.0) AS FLOAT64) AS avg_ep_dsr,
  CAST(COALESCE(p.cc_app_count, 0) AS INT64) AS cc_app_count,
  CAST(COALESCE(p.total_cc_limit_myr, 0) AS FLOAT64) AS total_cc_limit_myr,
  CAST(COALESCE(p.latest_ctos_score, 650) AS FLOAT64) AS latest_ctos_score,
  CAST(COALESCE(p.active_card_count, 0) AS INT64) AS active_card_count,
  CAST(COALESCE(p.total_cp_usage_myr, 0) AS FLOAT64) AS total_cp_usage_myr,
  CAST(COALESCE(p.total_cp_available_myr, 0) AS FLOAT64) AS total_cp_available_myr,
  ROUND(
    SAFE_DIVIDE(
      CAST(COALESCE(p.total_cp_usage_myr, 0) AS FLOAT64),
      NULLIF(CAST(COALESCE(p.total_cp_usage_myr, 0) + COALESCE(p.total_cp_available_myr, 0) AS FLOAT64), 0)
    ),
    4
  ) AS credit_utilization_ratio,
  CAST(COALESCE(p.combined_unpaid_osp, 0) AS FLOAT64) AS combined_unpaid_osp,
  COALESCE(p.worst_collection_score_grade, 'NONE') AS worst_collection_score_grade,
  CAST(COALESCE(cg.delinquent_exposed_merchants, 0) AS INT64) AS delinquent_exposed_merchants,
  CAST(COALESCE(cg.shared_delinquent_peers_count, 0) AS INT64) AS shared_delinquent_peers_count,
  CAST(COALESCE(cg.total_graph_merchant_spend_myr, 0) AS FLOAT64) AS total_graph_merchant_spend_myr,
  IF(COALESCE(p.combined_unpaid_osp, 0) > 0, 1, 0) AS label_is_delinquent,
  IF(COALESCE(p.active_card_count, 0) > 0, 1, 0) AS label_has_credit_card
FROM `acsm_gold.gold_aeon_customer360_profile` p
LEFT JOIN customer_graph_contagion cg
  ON p.CIF_ID = cg.cif_id;
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 1.2: Audit `acsm_gold.ml_customer_feature_store` Class Balance & Graph Contagion Features
SELECT
  COUNT(*) AS total_customers,
  COUNTIF(label_is_delinquent = 1) AS delinquent_label_count,
  ROUND(100.0 * COUNTIF(label_is_delinquent = 1) / COUNT(*), 2) AS delinquency_rate_pct,
  COUNTIF(label_has_credit_card = 1) AS credit_cardholder_count,
  COUNTIF(label_has_credit_card = 0 AND ep_app_count > 0 AND label_is_delinquent = 0) AS prime_ep_to_cc_cross_sell_pool,
  COUNTIF(delinquent_exposed_merchants > 0) AS customers_with_graph_merchant_contagion,
  ROUND(AVG(avg_ep_dsr), 2) AS avg_portfolio_dsr_pct,
  ROUND(AVG(latest_ctos_score), 1) AS avg_portfolio_ctos_score
FROM `acsm_gold.ml_customer_feature_store`;
````

## Step 2 (Backup Code — Module 4): Model Evaluation (`ML.EVALUATE`) & Explainable AI (`ML.GLOBAL_EXPLAIN` & `ML.EXPLAIN_PREDICT` Local SHAP Attributions)
**ACSM RFP Clauses**: `C1.1.4.7` *(Experiment Tracking & Evaluation)*, `C1.1.4.9` *(Transparent & Explainable AI)*, `C1.1.5.1` *(BNM RMiT Auditability)*

Under **Bank Negara Malaysia (BNM) RMiT** and **Responsible Financing** guidelines, credit control and underwriting models cannot operate as unexplainable black boxes. **Module 4** demonstrates:
1. **Model Evaluation (`ML.EVALUATE`) & Global Feature Importance (`ML.GLOBAL_EXPLAIN`)**: Audits `ROC AUC`, `precision`, `recall`, `f1_score`, and global feature weights across `model_delinquency_propensity` and `model_ep_to_cc_cross_sell`.
2. **Local SHAP Feature Attributions (`ML.EXPLAIN_PREDICT`)**: Computes the **Top 3 Local Feature Attributions (`STRUCT(3 AS top_k_features)`)** for every single customer (`100,000` rows) and materializes `acsm_gold.ml_customer_risk_and_cross_sell_scores` so Credit Control officers and `e-Buddy` know *why* a customer was flagged (e.g., `avg_ep_dsr`, `latest_ctos_score`, `credit_utilization_ratio`, or `shared_delinquent_peers_count`).

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 4.1: Evaluate Both Supervised Models (`ML.EVALUATE`) & Global Feature Importance (`ML.GLOBAL_EXPLAIN`)
SELECT
  '1. Credit Delinquency & AKPK Propensity (model_delinquency_propensity)' AS model_name,
  ROUND(precision, 4) AS precision,
  ROUND(recall, 4) AS recall,
  ROUND(accuracy, 4) AS accuracy,
  ROUND(f1_score, 4) AS f1_score,
  ROUND(log_loss, 4) AS log_loss,
  ROUND(roc_auc, 4) AS roc_auc
FROM ML.EVALUATE(MODEL `acsm_gold.model_delinquency_propensity`)
UNION ALL
SELECT
  '2. EP -> CC Cross-Sell Propensity (model_ep_to_cc_cross_sell)' AS model_name,
  ROUND(precision, 4) AS precision,
  ROUND(recall, 4) AS recall,
  ROUND(accuracy, 4) AS accuracy,
  ROUND(f1_score, 4) AS f1_score,
  ROUND(log_loss, 4) AS log_loss,
  ROUND(roc_auc, 4) AS roc_auc
FROM ML.EVALUATE(MODEL `acsm_gold.model_ep_to_cc_cross_sell`);
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 4.1b: Inspect Global Feature Importance (`ML.GLOBAL_EXPLAIN`) for Delinquency Propensity
SELECT
  feature,
  ROUND(attribution, 4) AS global_shap_importance
FROM ML.GLOBAL_EXPLAIN(MODEL `acsm_gold.model_delinquency_propensity`)
ORDER BY attribution DESC
LIMIT 10;
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 4.2: Materialize Batch Explainable Predictions (`ML.EXPLAIN_PREDICT`) & Cross-Sell Scores
-- Stores per-customer Delinquency Probability + Top 3 Local SHAP Reasons + EP->CC Cross-Sell Probability
CREATE OR REPLACE TABLE `acsm_gold.ml_customer_risk_and_cross_sell_scores`
CLUSTER BY State, CIF_ID
OPTIONS (
  description = 'Pre-computed Explainable ML Delinquency Risk & EP-to-CC Cross-Sell Propensity scores with Top 3 Local SHAP feature attributions per customer (Clauses C1.1.4.4, C1.1.4.9, M1.5.1).'
) AS
WITH delinq_explain AS (
  SELECT
    CIF_ID,
    CIF_NM,
    State,
    Region,
    Occupation,
    N_Age,
    B_NetIncome,
    avg_ep_dsr,
    latest_ctos_score,
    total_ep_financed_myr,
    total_cc_limit_myr,
    active_card_count,
    credit_utilization_ratio,
    combined_unpaid_osp,
    worst_collection_score_grade,
    RecvPromo_FG,
    delinquent_exposed_merchants,
    shared_delinquent_peers_count,
    label_is_delinquent,
    predicted_label_is_delinquent AS predicted_delinquency_flag,
    ROUND(probability, 4) AS delinquency_risk_probability,
    top_feature_attributions,
    CONCAT(
      top_feature_attributions[SAFE_OFFSET(0)].feature, ' (',
      CAST(ROUND(top_feature_attributions[SAFE_OFFSET(0)].attribution, 3) AS STRING), ')'
    ) AS top_1_shap_reason,
    CONCAT(
      top_feature_attributions[SAFE_OFFSET(1)].feature, ' (',
      CAST(ROUND(top_feature_attributions[SAFE_OFFSET(1)].attribution, 3) AS STRING), ')'
    ) AS top_2_shap_reason,
    CONCAT(
      top_feature_attributions[SAFE_OFFSET(2)].feature, ' (',
      CAST(ROUND(top_feature_attributions[SAFE_OFFSET(2)].attribution, 3) AS STRING), ')'
    ) AS top_3_shap_reason
  FROM ML.EXPLAIN_PREDICT(
    MODEL `acsm_gold.model_delinquency_propensity`,
    TABLE `acsm_gold.ml_customer_feature_store`,
    STRUCT(3 AS top_k_features)
  )
),
cross_sell_pred AS (
  SELECT
    CIF_ID,
    predicted_label_has_credit_card AS predicted_cc_cross_sell_flag,
    ROUND(
      (SELECT p.prob FROM UNNEST(predicted_label_has_credit_card_probs) p WHERE p.label = 1),
      4
    ) AS cc_cross_sell_probability
  FROM ML.PREDICT(
    MODEL `acsm_gold.model_ep_to_cc_cross_sell`,
    TABLE `acsm_gold.ml_customer_feature_store`
  )
)
SELECT
  d.*,
  cs.predicted_cc_cross_sell_flag,
  cs.cc_cross_sell_probability
FROM delinq_explain d
LEFT JOIN cross_sell_pred cs
  USING (CIF_ID);
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 4.3: Top 10 High-Risk Customers Flagged for Proactive AKPK Restructuring with Local SHAP Reasons
SELECT
  CIF_ID,
  State,
  B_NetIncome,
  avg_ep_dsr,
  latest_ctos_score,
  combined_unpaid_osp,
  worst_collection_score_grade,
  delinquent_exposed_merchants,
  delinquency_risk_probability,
  top_1_shap_reason,
  top_2_shap_reason,
  top_3_shap_reason
FROM `acsm_gold.ml_customer_risk_and_cross_sell_scores`
WHERE predicted_delinquency_flag = 1
ORDER BY delinquency_risk_probability DESC, combined_unpaid_osp DESC
LIMIT 10;
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 4.4: Top 10 Pre-Qualified Easy Payment (EP)-Only Customers for AEON Credit Card Cross-Sell
-- Enforces PDPA marketing consent (`RecvPromo_FG = 'Y'`), zero delinquency (`label_is_delinquent = 0`), and `active_card_count = 0`
SELECT
  CIF_ID,
  State,
  Occupation,
  B_NetIncome,
  ep_app_count,
  total_ep_financed_myr,
  avg_ep_dsr,
  latest_ctos_score,
  RecvPromo_FG AS pdpa_consent,
  cc_cross_sell_probability,
  CASE
    WHEN B_NetIncome >= 8000 AND latest_ctos_score >= 700 THEN 'AEON Platinum Visa (Fast-Track)'
    WHEN B_NetIncome >= 4000 AND latest_ctos_score >= 660 THEN 'AEON Gold Visa (Fast-Track)'
    ELSE 'AEON Classic Rewards Card'
  END AS recommended_card_offer
FROM `acsm_gold.ml_customer_risk_and_cross_sell_scores`
WHERE active_card_count = 0
  AND ep_app_count > 0
  AND label_is_delinquent = 0
  AND avg_ep_dsr <= 50.0
  AND RecvPromo_FG = 'Y'
ORDER BY cc_cross_sell_probability DESC, B_NetIncome DESC
LIMIT 10;
````

````python
%%bigquery --project $PROJECT_ID --location $LOCATION
-- Backup Module 5.2 (Capstone): Single-Query In-Warehouse SQL Graph RAG
-- Joins: (1) BQML SHAP Risk Reasons + (2) 2-Hop Property Graph Facility/Merchant Context
--      + (3) VECTOR_SEARCH BNM/AKPK Policy Clause -> (4) ML.GENERATE_TEXT (Gemini 2.5 Flash) Action Plan
WITH top_risk_customers AS (
  SELECT
    CIF_ID,
    State,
    B_NetIncome,
    avg_ep_dsr,
    latest_ctos_score,
    combined_unpaid_osp,
    worst_collection_score_grade,
    delinquent_exposed_merchants,
    delinquency_risk_probability,
    CONCAT(top_1_shap_reason, '; ', top_2_shap_reason, '; ', top_3_shap_reason) AS shap_risk_drivers
  FROM `acsm_gold.ml_customer_risk_and_cross_sell_scores`
  WHERE predicted_delinquency_flag = 1
  ORDER BY delinquency_risk_probability DESC, combined_unpaid_osp DESC
  LIMIT 5
),
graph_2hop_context AS (
  SELECT
    g.cif_id AS CIF_ID,
    STRING_AGG(DISTINCT CONCAT(g.product_line, ':', g.facility_id, ' (RM ', CAST(g.exposure_amount_myr AS STRING), ')'), ', ' LIMIT 3) AS graph_facilities_held,
    STRING_AGG(DISTINCT CONCAT(g.merchant_name, ' [', g.merchant_group, ']'), ', ' LIMIT 3) AS graph_merchants_transacted,
    ROUND(SUM(g.total_spend_myr), 2) AS graph_2hop_spend_myr
  FROM GRAPH_TABLE(
    `acsm_gold.acsm_credit_ecosystem_graph`
    MATCH (f:CreditFacility)<-[h:HOLDS_FACILITY]-(c:Customer)-[t:TRANSACTED_AT]->(m:Merchant)
    COLUMNS (
      c.cif_id AS cif_id,
      f.facility_id AS facility_id,
      f.product_line AS product_line,
      f.exposure_amount_myr AS exposure_amount_myr,
      m.merchant_name AS merchant_name,
      m.merchant_group AS merchant_group,
      t.total_spend_myr AS total_spend_myr
    )
  ) g
  INNER JOIN top_risk_customers r
    ON g.cif_id = r.CIF_ID
  GROUP BY g.cif_id
),
customer_query_embeddings AS (
  SELECT
    *
  FROM ML.GENERATE_EMBEDDING(
    MODEL `acsm_gold.model_vertex_text_embedding`,
    (
      SELECT
        r.CIF_ID,
        r.State,
        r.B_NetIncome,
        r.avg_ep_dsr,
        r.combined_unpaid_osp,
        r.worst_collection_score_grade,
        r.delinquent_exposed_merchants,
        r.delinquency_risk_probability,
        r.shap_risk_drivers,
        COALESCE(gc.graph_facilities_held, 'EP/CC Standard Facility') AS graph_facilities_held,
        COALESCE(gc.graph_merchants_transacted, 'General Retail') AS graph_merchants_transacted,
        CONCAT(
          'High delinquency propensity customer with DSR ', CAST(r.avg_ep_dsr AS STRING),
          '%, Unpaid OSP RM ', CAST(r.combined_unpaid_osp AS STRING),
          ', Collection Grade ', r.worst_collection_score_grade,
          ', and Graph delinquent merchant exposure count ', CAST(r.delinquent_exposed_merchants AS STRING),
          ' seeking AKPK debt restructuring and BNM RMiT affordability review.'
        ) AS content
      FROM top_risk_customers r
      LEFT JOIN graph_2hop_context gc
        USING (CIF_ID)
    ),
    STRUCT(TRUE AS flatten_json_output, 'RETRIEVAL_QUERY' AS task_type)
  )
),
vector_matched_context AS (
  SELECT
    vs.query.CIF_ID,
    vs.query.State,
    vs.query.B_NetIncome,
    vs.query.avg_ep_dsr,
    vs.query.combined_unpaid_osp,
    vs.query.delinquency_risk_probability,
    vs.query.shap_risk_drivers,
    vs.query.graph_facilities_held,
    vs.query.graph_merchants_transacted,
    vs.base.policy_id AS matched_policy_id,
    vs.base.policy_title AS matched_policy_title,
    ROUND(vs.distance, 4) AS cosine_distance,
    CONCAT(
      'You are an AEON Credit Service Malaysia (ACSM) Credit Control & Compliance Advisor in Singapore/Malaysia. ',
      'Write a concise 2-sentence BNM RMiT & AKPK Restructuring Action Plan for Customer CIF ', vs.query.CIF_ID,
      ' (State: ', vs.query.State, ', Net Income: RM ', CAST(vs.query.B_NetIncome AS STRING),
      ', DSR: ', CAST(vs.query.avg_ep_dsr AS STRING), '%, Unpaid OSP: RM ', CAST(vs.query.combined_unpaid_osp AS STRING), '). ',
      '1) BQML Delinquency Risk Probability: ', CAST(vs.query.delinquency_risk_probability AS STRING),
      ' with Top 3 Local SHAP Drivers: [', vs.query.shap_risk_drivers, ']. ',
      '2) BigQuery Property Graph 2-Hop Exposure: Facilities=[', vs.query.graph_facilities_held,
      '], Merchants=[', vs.query.graph_merchants_transacted, ']. ',
      '3) Retrieved Policy Clause (', vs.base.policy_id, ' - ', vs.base.policy_title, '): ', vs.base.content
    ) AS prompt
  FROM VECTOR_SEARCH(
    TABLE `acsm_gold.bnm_rmit_akpk_policy_embeddings`,
    'embedding',
    TABLE customer_query_embeddings,
    'ml_generate_embedding_result',
    top_k => 1,
    distance_type => 'COSINE'
  ) vs
)
SELECT
  CIF_ID,
  State,
  delinquency_risk_probability,
  shap_risk_drivers,
  graph_facilities_held,
  graph_merchants_transacted,
  matched_policy_id,
  cosine_distance,
  TRIM(ml_generate_text_llm_result) AS gemini_graph_rag_action_plan
FROM ML.GENERATE_TEXT(
  MODEL `acsm_gold.model_gemini_flash`,
  TABLE vector_matched_context,
  STRUCT(
    0.2 AS temperature,
    300 AS max_output_tokens,
    TRUE AS flatten_json_output
  )
)
ORDER BY delinquency_risk_probability DESC;
````
